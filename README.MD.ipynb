# 🧠 CIFAR-10 Image Classification using Deep CNN (TensorFlow)

Achieving **~90.5% test accuracy** on CIFAR-10 using a custom deep Convolutional Neural Network with Batch Normalization, Dropout, Data Augmentation, and Learning Rate Scheduling.

---

## 📌 Project Overview

This project implements a high-performance Convolutional Neural Network (CNN) using TensorFlow/Keras to classify images from the CIFAR-10 dataset into 10 distinct categories.

The model combines multiple deep learning techniques to improve convergence and generalization, achieving strong performance without using transfer learning.

### 📊 Results Summary

| Metric | Value |
|-------|------|
| ✅ Test Accuracy | **90.51%** |
| ✅ Test Loss | 0.3306 |
| 🧠 Total Parameters | ~3.25M |
| ⏱️ Training Epochs | 50 |

## 📂 Dataset: CIFAR-10

The CIFAR-10 dataset contains **60,000 32×32 color images** across 10 classes:

| Label | Class | Label | Class |
|------|------|------|------|
| 0 | Airplane ✈️ | 5 | Dog 🐶 |
| 1 | Automobile 🚗 | 6 | Frog 🐸 |
| 2 | Bird 🐦 | 7 | Horse 🐴 |
| 3 | Cat 🐱 | 8 | Ship 🚢 |
| 4 | Deer 🦌 | 9 | Truck 🚚 |

- Training set: 50,000 images  
- Test set: 10,000 images  

**Challenge:** Small image size (32×32) and high similarity between classes (e.g., cats vs dogs) make this a non-trivial benchmark.

---

## ⚙️ Model Architecture (Deep CNN)

A custom deep CNN with **3 convolutional blocks** followed by a dense classifier.

### 🔹 Convolutional Base

| Block | Layers |
|------|--------|
| 1 | Conv2D(64) → BN → Conv2D(64) → BN → MaxPool → Dropout(0.25) |
| 2 | Conv2D(128) → BN → Conv2D(128) → BN → MaxPool → Dropout(0.25) |
| 3 | Conv2D(256) → BN → Conv2D(256) → BN → MaxPool → Dropout(0.25) |

### 🔹 Fully Connected Head

- Flatten  
- Dense(512) + BatchNormalization + Dropout(0.5)  
- Dense(10, activation='softmax')  

**Total Parameters:** ~3.25 Million

## 🚀 Key Features & Techniques

| Technique | Purpose |
|----------|--------|
| ✅ Batch Normalization | Stabilizes training & speeds up convergence |
| ✅ Dropout (0.25–0.5) | Prevents overfitting |
| ✅ Data Augmentation | Rotation, shifts, horizontal flip |
| ✅ Learning Rate Scheduler | Reduces LR after epochs 20 & 30 |
| ✅ Adam Optimizer | Adaptive optimization |
| ✅ Categorical Crossentropy | Multi-class loss function |
